Q6. What is the main difference between a tuple and a list
Answer:
       The main difference between the list and tuple is that list are mutable and we can change it elements while tuple are immutable and it elements 
can not be change after the creation.
    
Syntax:
    List: Defined with []
    Tuple: Defined with ()
Performance
    Lists: Slightly slower due to mutability and dynamic resizing.
    Tuples: Faster to iterate and more memory‑efficient.
        
Use Cases:
    Lists: Best for data that changes frequently.
    example: dynamic collections.
    Tuples: Ideal for fixed data, constants, or when using as dictionary keys (tuples can be hashable).

Q7. Function: grab the domain from "user@domain.com".

```python
def grabdomain (x):
    return x.split('@')[1]
print(grabdomain("user@domain.com"))
```

Q9. Function: count how many times "dog" occurs in a string (case-insensitive).

Q10. caught_speeding(speed, is_birthday):≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    allowance=5 if is_birthday else 0
    
    if speed <= 60 + allowance:
        return "No ticket"
    elif speed <=80 + allowance:
        return "Small ticket"
    else:
        return "Big ticket"

print(caught_speeding(60,False))
print(caught_speeding(65,True))
print(caught_speeding(75,True))
print(caught_speeding(72,False))
print(caught_speeding(95,True))
```
